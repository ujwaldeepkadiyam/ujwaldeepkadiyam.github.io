# 📝 Assignment: Star Schema in Pandas

In a star schema, you have:

* **Fact table** (large, numeric measures → e.g., sales transactions)
* **Dimension tables** (smaller lookup tables with descriptive attributes)

We’ll create:

* **FactSales** (transactions)
* **DimCustomer**
* **DimProduct**
* **DimDate**
* **DimStore**

Then we’ll merge them together to form the star schema.



- **Objective**
Understand how to create a **star schema** with one **fact table** and multiple **dimension tables**, then join them to form a denormalized dataset.

- Create Dimension Tables



````python
import pandas as pd

# Customer Dimension
DimCustomer = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4, 5],
    "CustomerName": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "Region": ["North", "South", "East", "West", "North"]
})

# Product Dimension
DimProduct = pd.DataFrame({
    "ProductID": [101, 102, 103],
    "ProductName": ["Laptop", "Phone", "Tablet"],
    "Category": ["Electronics", "Electronics", "Electronics"]
})

# Date Dimension
DimDate = pd.DataFrame({
    "DateID": [20230101, 20230102, 20230103],
    "Date": ["2023-01-01", "2023-01-02", "2023-01-03"],
    "Day": ["Sunday", "Monday", "Tuesday"]
})

# Store Dimension
DimStore = pd.DataFrame({
    "StoreID": [11, 12],
    "StoreName": ["Downtown", "Mall"],
    "City": ["New York", "Boston"]
})

DimCustomer, DimProduct, DimDate, DimStore
````

-  Create Fact Table

````python
# Fact Table: Sales Transactions
FactSales = pd.DataFrame({
    "SaleID": range(1, 11),
    "CustomerID": [1, 2, 3, 1, 5, 4, 2, 3, 5, 1],
    "ProductID": [101, 102, 103, 101, 102, 103, 101, 102, 103, 101],
    "DateID": [20230101, 20230102, 20230103, 20230101, 20230102, 20230103, 20230101, 20230102, 20230103, 20230101],
    "StoreID": [11, 12, 11, 11, 12, 11, 12, 12, 11, 11],
    "Quantity": [2, 1, 3, 1, 2, 1, 4, 2, 5, 1],
    "Revenue": [2000, 500, 1500, 1000, 1000, 600, 4000, 1000, 2500, 900]
})

FactSales
````

### Question:

Join the fact table with all the dimension tables to form a **denormalized star schema dataset**.


### Solution-1

### Solution - 2

```python
star_schema = FactSales \
    .merge(DimCustomer, on="CustomerID", how="left") \
    .merge(DimProduct, on="ProductID", how="left") \
    .merge(DimDate, on="DateID", how="left") \
    .merge(DimStore, on="StoreID", how="left")

star_schema
```

## 🧠 Explanation

* The **FactSales** table stores the **measures** (Quantity, Revenue).
* The **dimension tables** add descriptive context:

  * **DimCustomer** → who bought
  * **DimProduct** → what was bought
  * **DimDate** → when it was bought
  * **DimStore** → where it was bought
* The star schema lets us run **analytics** like:

  * Total revenue by region
  * Quantity sold per product category
  * Sales per day or per store

# 📝 Assignment: Snowflake Schema in Pandas
The **snowflake schema** normalizes some of the dimension tables into **sub-dimensions** (extra lookup tables). For example:

- **Objective**
Understand how a **snowflake schema** extends a star schema by normalizing dimension tables into additional related dimensions.


- Create Dimension Tables

````python
import pandas as pd

# Customer Dimension (normalized: Region is now a foreign key)
DimCustomer = pd.DataFrame({
    "CustomerID": [1, 2, 3, 4, 5],
    "CustomerName": ["Alice", "Bob", "Charlie", "David", "Eva"],
    "RegionID": [101, 102, 101, 103, 102]
})

# Region Dimension (separate lookup table)
DimRegion = pd.DataFrame({
    "RegionID": [101, 102, 103],
    "RegionName": ["North", "South", "West"]
})

# Product Dimension
DimProduct = pd.DataFrame({
    "ProductID": [201, 202, 203],
    "ProductName": ["Laptop", "Phone", "Tablet"],
    "Category": ["Electronics", "Electronics", "Electronics"]
})

# Date Dimension
DimDate = pd.DataFrame({
    "DateID": [20230101, 20230102, 20230103],
    "Date": ["2023-01-01", "2023-01-02", "2023-01-03"],
    "Day": ["Sunday", "Monday", "Tuesday"]
})

# Store Dimension
DimStore = pd.DataFrame({
    "StoreID": [11, 12],
    "StoreName": ["Downtown", "Mall"],
    "City": ["New York", "Boston"]
})

# Fact Table
FactSales = pd.DataFrame({
    "SaleID": range(1, 6),
    "CustomerID": [1, 2, 3, 4, 5],
    "ProductID": [201, 202, 203, 201, 202],
    "DateID": [20230101, 20230102, 20230103, 20230101, 20230102],
    "StoreID": [11, 12, 11, 12, 11],
    "Quantity": [2, 1, 3, 4, 2],
    "Revenue": [2000, 500, 1500, 4000, 1000]
})
````

| Table Name      | Type of Table | Primary Key | Present In (as Foreign Key) |
| --------------- | ------------- | ----------- | --------------------------- |
| **FactSales**   | Fact          | SaleID      | —                           |
| **DimCustomer** | Dimension     | CustomerID  | FactSales                   |
| **DimRegion**   | Sub-dimension | RegionID    | DimCustomer                 |
| **DimProduct**  | Dimension     | ProductID   | FactSales                   |
| **DimCategory** | Sub-dimension | CategoryID  | DimProduct                  |
| **DimDate**     | Dimension     | DateID      | FactSales                   |
| **DimStore**    | Dimension     | StoreID     | FactSales                   |
| **DimCity**     | Sub-dimension | CityID      | DimStore                    |
