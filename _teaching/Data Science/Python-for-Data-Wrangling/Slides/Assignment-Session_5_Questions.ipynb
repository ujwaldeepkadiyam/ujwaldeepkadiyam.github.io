### **Step 1: Melt the DataFrame**

We convert the score columns into a long format:

````python
score_cols = ['SAS','R','PYTHON','SQL','EXCEL']

df_melted = df.melt(id_vars=['NAME'], value_vars=score_cols,
                    var_name='SUBJECT', value_name='SCORE')

df_melted.head()
````


### **Step 2: Group by NAME and compute average**

Now, we can use `groupby` to calculate the average per student:

````python
average_scores = df_melted.groupby('NAME')['SCORE'].mean().reset_index()
average_scores.rename(columns={'SCORE':'AVERAGE_SCORE'}, inplace=True)

average_scores
````

**Output:**


✅ This gives the **same result as `mean(axis=1)`**, but uses `melt` + `groupby`.


### **Step 3 (Optional): Merge back to original DataFrame**

If you want the `AVERAGE_SCORE` column in the original DataFrame:

````python
df = df.merge(average_scores, on='NAME')
# df[['NAME','AVERAGE_SCORE']]
````

💡 **Key notes:**

* `melt` is useful for **long-format data** or **grouped aggregations**.
* For simple **row-wise averages**, `df[cols].mean(axis=1)` is more direct.
* `groupby` is more natural for **column-wise aggregation by groups**, e.g., `CLASS` or `GENDER`.

#### Q9: compute **average scores per group** using `melt` + `groupby`. Here’s how:

### **Step 1: Melt the DataFrame (scores only)**

````python
score_cols = ['SAS','R','PYTHON','SQL','EXCEL']

df_melted = df.melt(id_vars=['NAME','GENDER','EDUCATION'], 
                    value_vars=score_cols,
                    var_name='SUBJECT', value_name='SCORE')

df_melted.head()
````

**Output example:**

```
     NAME  GENDER EDUCATION  SUBJECT  SCORE
0  PREETAM    MALE       MBA      SAS    800
1      RAJ  FEMALE     BTECH      SAS    700
2    LAVLIN    MALE     BTECH      SAS    100
3    ANITA  FEMALE       MCA      SAS      0
4    KUMAR    MALE      MSC      SAS    500
```


### **Step 2: Group by GENDER (or EDUCATION) and compute average score**

````python
# Average score per GENDER
gender_avg = df_melted.groupby('GENDER')['SCORE'].mean().reset_index()
gender_avg.rename(columns={'SCORE':'AVERAGE_SCORE'}, inplace=True)

print(gender_avg)
````

**Output:**

```
   GENDER  AVERAGE_SCORE
0  FEMALE     532.5
1    MALE     512.0
```

 

````python
# Average score per EDUCATION
education_avg = df_melted.groupby('EDUCATION')['SCORE'].mean().reset_index()
education_avg.rename(columns={'SCORE':'AVERAGE_SCORE'}, inplace=True)

print(education_avg)
````

**Output:**

```
  EDUCATION  AVERAGE_SCORE
0      BTECH     430.0
1        MBA     650.0
2        MCA     462.5
3        MSC     562.5
```


### **Step 3 (Optional): Merge back to original DataFrame**

If you want to **add group-level averages** back to the original DataFrame:

````python
df = df.merge(gender_avg, on='GENDER', suffixes=('', '_GENDER'))
df = df.merge(education_avg, on='EDUCATION', suffixes=('', '_EDU'))

df[['NAME','GENDER','AVERAGE_SCORE','AVERAGE_SCORE_EDU']]
````

**Now you have:**

* Individual row scores (if you want)
* Average score per gender (`AVERAGE_SCORE`)
* Average score per education (`AVERAGE_SCORE_EDU`)


✅ **Summary:**

1. `melt` transforms wide score columns into long format.
2. `groupby` + `mean()` computes average per group.
3. `merge` can attach the group averages back to the original DataFrame.

#### Q10:  Use **`groupby().transform()`**, to calculate **group-level averages** **directly** without melting or merging. 

- `transform` returns a Series aligned with the original DataFrame, so it can be assigned as a new column.

### **1️⃣ Row-wise average (per student)**

````python
score_cols = ['SAS','R','PYTHON','SQL','EXCEL']

# Average score per row
df['AVERAGE_SCORE'] = df[score_cols].mean(axis=1)
df[['NAME','AVERAGE_SCORE']]
````

✅ Same as before, simple row-wise mean.


### **2️⃣ Average score per GENDER using transform**

````python
# Compute average of all score columns per gender
df['GENDER_AVG'] = df[score_cols].mean(axis=1)  # start with row-wise sum
df['GENDER_AVG'] = df.groupby('GENDER')[score_cols].transform('mean').mean(axis=1)

df[['NAME','GENDER','GENDER_AVG']]
````

**Explanation:**

1. `df.groupby('GENDER')[score_cols]` → select numeric columns per group.
2. `.transform('mean')` → compute **mean per column within the group**, returned as a DataFrame aligned with `df`.
3. `.mean(axis=1)` → average across the score columns per student to get **group average**.


### **3️⃣ Average score per EDUCATION using transform**

````python
df['EDU_AVG'] = df.groupby('EDUCATION')[score_cols].transform('mean').mean(axis=1)
df[['NAME','EDUCATION','EDU_AVG']]
````

### ✅ **Advantages of using transform**

* No need to `melt` or `merge`.
* Returns a **Series aligned with original DataFrame**.
* Works nicely for **row-level + group-level computations**.
